# CNN + CE — 226 Hand-Specific Classes

This Colab notebook is the **matched CE-only baseline** for the 226-class hand-specific experiment.

It uses:
- 113 people × 2 hands = **226 classes**
- DB1 = **904 images** = 4 images/class
- DB2 = **678 images** = 3 images/class
- DB1 split per class: **3 train + 1 validation**
- DB2: all 3 images/class for cross-session testing
- Same six-layer CNN backbone and 256-D embedding as the ArcFace experiment
- **Cross-entropy supervision only**
- seeds: `42, 123, 456, 2026, 777`
- resumable checkpoints after every epoch
- the same final biometric evaluation: DB1 gallery centroids + DB2 probes + cosine similarity

For a clean ablation, keep all settings identical to the ArcFace notebook except for removing the ArcFace branch and angular-margin loss.

```python
SEEDS = [42, 123, 456, 2026, 777]
NUM_PEOPLE = 113
NUM_CLASSES = 226

IMG_SIZE = 224
BATCH_SIZE = 32
MAX_EPOCHS = 200
LR = 1e-3
WEIGHT_DECAY = 1e-4
EMBED_DIM = 256


NUM_WORKERS = 0
VALID_HANDS = {"L", "R"}

# Set this to the SAME filter/tag used in your 904/678 experiments
FILTER_TYPE = "roi"

def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False
```

```python
PATTERN = re.compile(
    r"^person_(\d{3})_db(\d)_([LR])(\d+)_([a-zA-Z]+)\.png$",
    re.IGNORECASE
)

def scan_dataset(root):
    rows = []
    for p in Path(root).rglob("*.png"):
        m = PATTERN.match(p.name)
        if m is None:
            continue

        person = int(m.group(1))
        db = int(m.group(2))
        hand = m.group(3).upper()
        sample = int(m.group(4))
        tag = m.group(5).lower()

        if not (1 <= person <= NUM_PEOPLE):
            continue
        if hand not in VALID_HANDS:
            continue
        if tag != FILTER_TYPE:
            continue

        rows.append({
            "path": str(p),
            "person": person,
            "db": db,
            "hand": hand,
            "sample": sample,
            "tag": tag
        })

    df = pd.DataFrame(rows)
    assert len(df) > 0, "No matching images. Check DATA_ROOT/FILTER_TYPE."
    return df.sort_values(["person","db","hand","sample"]).reset_index(drop=True)

df = scan_dataset(DATA_ROOT)
db1 = df[df.db == 1].copy().reset_index(drop=True)
db2 = df[df.db == 2].copy().reset_index(drop=True)

assert len(db1) == 904, f"Expected DB1=904, got {len(db1)}"
assert len(db2) == 678, f"Expected DB2=678, got {len(db2)}"

def hand_label(person, hand):
    return (person - 1) * 2 + (0 if hand == "L" else 1)

db1["label"] = [hand_label(p,h) for p,h in zip(db1.person, db1.hand)]
db2["label"] = [hand_label(p,h) for p,h in zip(db2.person, db2.hand)]

assert db1.label.nunique() == NUM_CLASSES
assert db2.label.nunique() == NUM_CLASSES
assert (db1.groupby("label").size() == 4).all()
assert (db2.groupby("label").size() == 3).all()

print("DB1:", len(db1), "| DB2:", len(db2), "| classes:", NUM_CLASSES)
```

```python
def split_db1(df):
    tr, va = [], []
    for label in sorted(df.label.unique()):
        c = df[df.label == label].sort_values("sample").reset_index(drop=True)
        assert len(c) == 4
        tr.append(c.iloc[:3])
        va.append(c.iloc[3:4])
    return pd.concat(tr, ignore_index=True), pd.concat(va, ignore_index=True)

train_df, val_df = split_db1(db1)
test_df = db2.copy().reset_index(drop=True)

assert len(train_df) == 678
assert len(val_df) == 226
assert len(test_df) == 678
print("Train:", len(train_df), "Val:", len(val_df), "Test:", len(test_df))
```

```python
def ac(img):
    return ImageOps.autocontrast(img.convert("L"), cutoff=2)

train_tf = transforms.Compose([
    transforms.Lambda(ac),
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.RandomRotation(3, fill=128),
    transforms.RandomAffine(0, translate=(0.02,0.02), scale=(0.98,1.02), fill=128),
    transforms.ColorJitter(brightness=0.05, contrast=0.05),
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))
])

eval_tf = transforms.Compose([
    transforms.Lambda(ac),
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))
])

class VeinDataset(Dataset):
    def __init__(self, frame, tf):
        self.frame = frame.reset_index(drop=True)
        self.tf = tf
    def __len__(self):
        return len(self.frame)
    def __getitem__(self, i):
        r = self.frame.iloc[i]
        with Image.open(r.path) as im:
            x = self.tf(im)
        return x, int(r.label)

train_ds = VeinDataset(train_df, train_tf)
val_ds = VeinDataset(val_df, eval_tf)
test_ds = VeinDataset(test_df, eval_tf)

def make_train_loader(seed, epoch):
    g = torch.Generator()
    g.manual_seed(seed * 100000 + epoch)
    return DataLoader(
        train_ds, batch_size=BATCH_SIZE, shuffle=True,
        drop_last=True, num_workers=NUM_WORKERS,
        pin_memory=torch.cuda.is_available(), generator=g
    )

def make_eval_loader(ds):
    return DataLoader(
        ds, batch_size=BATCH_SIZE, shuffle=False,
        drop_last=False, num_workers=NUM_WORKERS,
        pin_memory=torch.cuda.is_available()
    )

val_loader = make_eval_loader(val_ds)
test_loader = make_eval_loader(test_ds)
```

```python
class Block(nn.Module):
    def __init__(self, a, b, pool=True):
        super().__init__()
        layers = [
            nn.Conv2d(a,b,3,padding=1,bias=False),
            nn.BatchNorm2d(b),
            nn.ReLU(inplace=True)
        ]
        if pool:
            layers.append(nn.MaxPool2d(2))
        self.net = nn.Sequential(*layers)

    def forward(self, x):
        return self.net(x)


class Backbone(nn.Module):
    def __init__(self, d=256):
        super().__init__()
        self.f = nn.Sequential(
            Block(1,32), Block(32,64), Block(64,128), Block(128,256),
            Block(256,256,False), Block(256,512,False)
        )
        self.gap = nn.AdaptiveAvgPool2d(1)
        self.emb = nn.Sequential(
            nn.Dropout(0.5),
            nn.Linear(512,d),
            nn.BatchNorm1d(d),
            nn.ReLU(inplace=True),
            nn.Dropout(0.25)
        )

    def forward(self, x):
        return self.emb(self.gap(self.f(x)).flatten(1))


class CNNCE226(nn.Module):
    """Matched 226-class baseline with conventional cross-entropy only."""

    def __init__(self):
        super().__init__()
        self.backbone = Backbone(EMBED_DIM)
        self.ce = nn.Linear(EMBED_DIM, NUM_CLASSES)

    def forward(self, x, y=None):
        # y is accepted only to preserve the same calling convention
        # used by the ArcFace notebook.
        z = self.backbone(x)
        ce = self.ce(z)
        return z, ce, None
```

```python
def atomic_torch_save(obj, path):
    path = Path(path)
    tmp = path.with_suffix(path.suffix + ".tmp")
    torch.save(obj, tmp)
    os.replace(tmp, path)

def atomic_csv_save(df, path):
    path = Path(path)
    tmp = path.with_suffix(path.suffix + ".tmp")
    df.to_csv(tmp, index=False)
    os.replace(tmp, path)

def save_rng_state():
    state = {
        "python": random.getstate(),
        "numpy": np.random.get_state(),
        "torch": torch.get_rng_state().cpu()
    }
    if torch.cuda.is_available():
        state["cuda"] = [s.cpu() for s in torch.cuda.get_rng_state_all()]
    return state

def restore_rng_state(state):
    if not state:
        return
    random.setstate(state["python"])
    np.random.set_state(state["numpy"])
    torch.set_rng_state(state["torch"].detach().cpu().to(torch.uint8))
    if torch.cuda.is_available() and "cuda" in state and state["cuda"] is not None:
        cuda_states = [s.detach().cpu().to(torch.uint8) for s in state["cuda"]]
        torch.cuda.set_rng_state_all(cuda_states[:torch.cuda.device_count()])

def safe_torch_load(path, map_location="cpu"):
    try:
        return torch.load(path, map_location=map_location, weights_only=False)
    except TypeError:
        return torch.load(path, map_location=map_location)

def optimizer_to(optimizer, device):
    for state in optimizer.state.values():
        for k,v in state.items():
            if isinstance(v, torch.Tensor):
                state[k] = v.to(device)
```

```python
@torch.no_grad()
def evaluate(model, loader):
    model.eval()
    loss_sum = n = c1 = c5 = 0
    ys, ps = [], []

    for x,y in loader:
        x,y = x.to(DEVICE), y.to(DEVICE)
        _,logits,_ = model(x)
        loss = F.cross_entropy(logits,y)
        bs = len(y)
        loss_sum += loss.item()*bs
        n += bs

        top5 = logits.topk(5,1).indices
        pred = top5[:,0]
        c1 += pred.eq(y).sum().item()
        c5 += top5.eq(y[:,None]).any(1).sum().item()

        ys.extend(y.cpu().tolist())
        ps.extend(pred.cpu().tolist())

    return {
        "loss": loss_sum/n,
        "rank1": c1/n,
        "rank5": c5/n,
        "y_true": np.array(ys),
        "y_pred": np.array(ps)
    }
```

```python
def train_seed(seed):
    seed_dir = OUT_DIR / f"seed_{seed}"
    seed_dir.mkdir(parents=True, exist_ok=True)

    latest = seed_dir / "latest.pt"
    best = seed_dir / "best.pt"
    history_csv = seed_dir / "history.csv"
    result_json = seed_dir / "result.json"

    if result_json.exists():
        with open(result_json, "r") as f:
            print(f"Seed {seed} already complete.")
            return json.load(f)

    set_seed(seed)
    model = CNNCE226().to(DEVICE)
    opt = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)
    sch = torch.optim.lr_scheduler.CosineAnnealingLR(opt, T_max=MAX_EPOCHS)

    start_epoch = 1
    best_val_loss = float("inf")
    best_epoch = -1
    history = []

    if latest.exists():
        print(f"Resume checkpoint found for seed {seed}.")
        ckpt = safe_torch_load(latest, "cpu")

        model.load_state_dict(ckpt["model_state"])
        model.to(DEVICE)

        opt.load_state_dict(ckpt["optimizer_state"])
        optimizer_to(opt, DEVICE)

        sch.load_state_dict(ckpt["scheduler_state"])

        start_epoch = int(ckpt["epoch"]) + 1
        best_val_loss = float(ckpt["best_val_loss"])
        best_epoch = int(ckpt["best_epoch"])
        history = ckpt.get("history", [])

        try:
            restore_rng_state(ckpt.get("rng_state"))
            print("RNG restored.")
        except Exception as e:
            print("RNG restore warning:", e)

        print("Resuming from epoch", start_epoch)

    for epoch in range(start_epoch, MAX_EPOCHS+1):
        model.train()
        loader = make_train_loader(seed, epoch)

        total_loss = total = correct = 0

        for x,y in loader:
            x,y = x.to(DEVICE), y.to(DEVICE)

            opt.zero_grad(set_to_none=True)
            _,ce,_ = model(x,y)

            loss = F.cross_entropy(ce,y)
            loss.backward()
            opt.step()

            bs = len(y)
            total_loss += loss.item()*bs
            total += bs
            correct += ce.argmax(1).eq(y).sum().item()

        sch.step()

        tr_loss = total_loss/total
        tr_acc = correct/total
        val = evaluate(model, val_loader)

        improved = val["loss"] < best_val_loss

        if improved:
            best_val_loss = val["loss"]
            best_epoch = epoch
            atomic_torch_save({
                "epoch": epoch,
                "model_state": model.state_dict(),
                "best_val_loss": best_val_loss,
                "best_epoch": best_epoch
            }, best)

        history.append({
            "epoch": epoch,
            "train_loss": tr_loss,
            "train_acc": tr_acc,
            "val_loss": val["loss"],
            "val_acc": val["rank1"],
            "learning_rate": opt.param_groups[0]["lr"]
        })
        atomic_csv_save(pd.DataFrame(history), history_csv)

        atomic_torch_save({
            "epoch": epoch,
            "model_state": model.state_dict(),
            "optimizer_state": opt.state_dict(),
            "scheduler_state": sch.state_dict(),
            "best_val_loss": best_val_loss,
            "best_epoch": best_epoch,
            "history": history,
            "rng_state": save_rng_state()
        }, latest)

        print(
            f"seed {seed:4d} | {epoch:03d}/{MAX_EPOCHS} | "
            f"tr {tr_acc*100:6.2f}% | val {val['rank1']*100:6.2f}% | "
            f"val_loss {val['loss']:.6f}"
            + (" <-- BEST" if improved else "")
        )

    ckpt = safe_torch_load(best, "cpu")
    model.load_state_dict(ckpt["model_state"])
    model.to(DEVICE)

    final_val = evaluate(model, val_loader)
    final_test = evaluate(model, test_loader)

    mp,mr,mf,_ = precision_recall_fscore_support(
        final_test["y_true"], final_test["y_pred"],
        labels=np.arange(NUM_CLASSES), average="macro", zero_division=0
    )
    _,_,wf,_ = precision_recall_fscore_support(
        final_test["y_true"], final_test["y_pred"],
        labels=np.arange(NUM_CLASSES), average="weighted", zero_division=0
    )

    result = {
        "experiment": "CNN_CE_226",
        "seed": int(seed),
        "classes": NUM_CLASSES,
        "best_epoch": int(best_epoch),
        "db1_val_loss": float(final_val["loss"]),
        "db1_val_acc": float(final_val["rank1"]),
        "db2_rank1": float(final_test["rank1"]),
        "db2_rank5": float(final_test["rank5"]),
        "macro_precision": float(mp),
        "macro_recall": float(mr),
        "macro_f1": float(mf),
        "weighted_f1": float(wf)
    }

    np.savez(
        seed_dir/"db2_predictions.npz",
        y_true=final_test["y_true"],
        y_pred=final_test["y_pred"]
    )

    with open(result_json, "w") as f:
        json.dump(result, f, indent=2)

    del model
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    return result
```

```python
# ============================================================
# EMBEDDING EXTRACTION
# ============================================================

@torch.no_grad()
def extract_embeddings(
    model,
    loader
):
    model.eval()

    embeddings = []
    labels = []

    for images, y in loader:

        images = images.to(
            DEVICE,
            non_blocking=True
        )

        z, _, _ = model(images)

        # Normalize embedding so dot product = cosine similarity
        z = F.normalize(
            z,
            p=2,
            dim=1
        )

        embeddings.append(
            z.cpu()
        )

        labels.append(
            y.cpu()
        )

    embeddings = torch.cat(
        embeddings,
        dim=0
    ).numpy()

    labels = torch.cat(
        labels,
        dim=0
    ).numpy()

    return embeddings, labels
```

```python
# ============================================================
# GALLERY TEMPLATE CONSTRUCTION
# ============================================================

def build_centroids(
    embeddings,
    labels
):
    centroids = []

    for class_id in range(
        NUM_CLASSES
    ):

        class_embeddings = (
            embeddings[
                labels == class_id
            ]
        )

        assert (
            len(class_embeddings) == 4
        ), (
            f"Expected 4 DB1 samples "
            f"for class {class_id}"
        )

        centroid = (
            class_embeddings.mean(
                axis=0
            )
        )

        centroid /= (
            np.linalg.norm(
                centroid
            )
            + 1e-12
        )

        centroids.append(
            centroid
        )

    return np.stack(
        centroids,
        axis=0
    )
```

```python
# ============================================================
# CLOSED-SET IDENTIFICATION / CMC
# ============================================================

def identification_metrics(
    probe_embeddings,
    probe_labels,
    centroids,
    max_rank=20
):

    # normalized vectors:
    # dot product = cosine similarity
    scores = (
        probe_embeddings
        @ centroids.T
    )

    ranking = np.argsort(
        -scores,
        axis=1
    )

    true_ranks = []

    for i, true_label in enumerate(
        probe_labels
    ):

        rank = (
            np.where(
                ranking[i]
                == true_label
            )[0][0]
            + 1
        )

        true_ranks.append(
            rank
        )

    true_ranks = np.asarray(
        true_ranks
    )

    cmc = np.array([
        np.mean(
            true_ranks <= k
        )
        for k in range(
            1,
            max_rank + 1
        )
    ])

    predictions = ranking[:, 0]

    return {
        "scores": scores,
        "ranking": ranking,
        "ranks": true_ranks,
        "predictions": predictions,
        "cmc": cmc,

        "rank1":
            np.mean(
                true_ranks <= 1
            ),

        "rank5":
            np.mean(
                true_ranks <= 5
            ),

        "rank10":
            np.mean(
                true_ranks <= 10
            )
    }
```

```python
# ============================================================
# VERIFICATION SCORES
# ============================================================

def generate_verification_scores(
    score_matrix,
    probe_labels
):

    genuine = []
    impostor = []

    for i, true_class in enumerate(
        probe_labels
    ):

        genuine.append(
            score_matrix[
                i,
                true_class
            ]
        )

        mask = (
            np.arange(
                NUM_CLASSES
            )
            != true_class
        )

        impostor.extend(
            score_matrix[
                i,
                mask
            ]
        )

    return (
        np.asarray(genuine),
        np.asarray(impostor)
    )
```

```python
# ============================================================
# ROC / AUC / EER
# ============================================================

def verification_metrics(
    genuine_scores,
    impostor_scores
):

    y_true = np.concatenate([
        np.ones(
            len(genuine_scores)
        ),
        np.zeros(
            len(impostor_scores)
        )
    ])

    y_score = np.concatenate([
        genuine_scores,
        impostor_scores
    ])

    fpr, tpr, thresholds = roc_curve(
        y_true,
        y_score
    )

    fnr = 1.0 - tpr

    eer_index = np.argmin(
        np.abs(
            fpr - fnr
        )
    )

    eer = (
        fpr[eer_index]
        + fnr[eer_index]
    ) / 2.0

    eer_threshold = (
        thresholds[
            eer_index
        ]
    )

    auc_value = roc_auc_score(
        y_true,
        y_score
    )

    return {
        "fpr": fpr,
        "tpr": tpr,
        "fnr": fnr,
        "thresholds": thresholds,

        "auc": auc_value,

        "eer": eer,

        "eer_threshold":
            eer_threshold,

        "far_at_eer":
            fpr[eer_index],

        "frr_at_eer":
            fnr[eer_index]
    }
```

```python
# ============================================================
# TAR @ FIXED FAR
# ============================================================

def tar_at_far(
    fpr,
    tpr,
    target_far
):

    valid = np.where(
        fpr <= target_far
    )[0]

    if len(valid) == 0:
        return np.nan

    return np.max(
        tpr[valid]
    )
```

```python
# ============================================================
# HAND-IDENTITY BOOTSTRAP 95% CI
# ============================================================

def bootstrap_rank1_ci(
    labels,
    predictions,
    iterations=2000,
    seed=2026
):

    rng = np.random.default_rng(
        seed
    )

    identities = np.unique(
        labels
    )

    boot_scores = []

    for _ in range(
        iterations
    ):

        sampled = rng.choice(
            identities,
            size=len(identities),
            replace=True
        )

        identity_acc = []

        for identity in sampled:

            mask = (
                labels == identity
            )

            accuracy = np.mean(
                predictions[mask]
                == labels[mask]
            )

            identity_acc.append(
                accuracy
            )

        boot_scores.append(
            np.mean(
                identity_acc
            )
        )

    boot_scores = np.asarray(
        boot_scores
    )

    return {
        "mean":
            boot_scores.mean(),

        "lower":
            np.percentile(
                boot_scores,
                2.5
            ),

        "upper":
            np.percentile(
                boot_scores,
                97.5
            ),

        "samples":
            boot_scores
    }
```

```python
# ============================================================
# EVALUATE ONE SAVED SEED
# ============================================================

def evaluate_biometric_seed(seed):

    print(
        "\\n" + "=" * 80
    )

    print(
        f"BIOMETRIC EVALUATION "
        f"— SEED {seed}"
    )

    print(
        "=" * 80
    )

    seed_dir = (
        OUT_DIR /
        f"seed_{seed}"
    )

    best_path = (
        seed_dir /
        "best.pt"
    )

    assert best_path.exists()

    # -----------------------------
    # Load best model
    # -----------------------------

    model = (
        CNNCE226()
        .to(DEVICE)
    )

    checkpoint = safe_torch_load(
        best_path,
        map_location="cpu"
    )

    model.load_state_dict(
        checkpoint[
            "model_state"
        ]
    )

    model.to(DEVICE)
    model.eval()

    # -----------------------------
    # Embeddings
    # -----------------------------

    gallery_z, gallery_y = (
        extract_embeddings(
            model,
            gallery_loader
        )
    )

    probe_z, probe_y = (
        extract_embeddings(
            model,
            probe_loader
        )
    )

    assert len(gallery_z) == 904
    assert len(probe_z) == 678

    # -----------------------------
    # Gallery templates
    # -----------------------------

    centroids = build_centroids(
        gallery_z,
        gallery_y
    )

    # -----------------------------
    # Identification
    # -----------------------------

    identification = (
        identification_metrics(
            probe_z,
            probe_y,
            centroids,
            max_rank=CMC_MAX_RANK
        )
    )

    # -----------------------------
    # Verification
    # -----------------------------

    genuine, impostor = (
        generate_verification_scores(
            identification[
                "scores"
            ],
            probe_y
        )
    )

    verification = (
        verification_metrics(
            genuine,
            impostor
        )
    )

    # -----------------------------
    # TAR @ FAR
    # -----------------------------

    tar_1 = tar_at_far(
        verification["fpr"],
        verification["tpr"],
        0.01
    )

    tar_01 = tar_at_far(
        verification["fpr"],
        verification["tpr"],
        0.001
    )

    # -----------------------------
    # Bootstrap
    # -----------------------------

    ci = bootstrap_rank1_ci(
        probe_y,
        identification[
            "predictions"
        ],
        iterations=
            BOOTSTRAP_ITERATIONS,
        seed=
            BOOTSTRAP_SEED
            + seed
    )

    # -----------------------------
    # Save
    # -----------------------------

    save_dir = (
        BIO_DIR /
        f"seed_{seed}"
    )

    save_dir.mkdir(
        parents=True,
        exist_ok=True
    )

    np.savez(
        save_dir /
        "biometric_scores.npz",

        probe_labels=
            probe_y,

        predictions=
            identification[
                "predictions"
            ],

        ranks=
            identification[
                "ranks"
            ],

        similarity=
            identification[
                "scores"
            ],

        genuine=
            genuine,

        impostor=
            impostor,

        cmc=
            identification[
                "cmc"
            ],

        fpr=
            verification[
                "fpr"
            ],

        tpr=
            verification[
                "tpr"
            ]
    )

    result = {

        "seed":
            int(seed),

        "rank1":
            float(
                identification[
                    "rank1"
                ]
            ),

        "rank5":
            float(
                identification[
                    "rank5"
                ]
            ),

        "rank10":
            float(
                identification[
                    "rank10"
                ]
            ),

        "roc_auc":
            float(
                verification[
                    "auc"
                ]
            ),

        "eer":
            float(
                verification[
                    "eer"
                ]
            ),

        "eer_threshold":
            float(
                verification[
                    "eer_threshold"
                ]
            ),

        "tar_far_1pct":
            float(tar_1),

        "tar_far_0_1pct":
            float(tar_01),

        "rank1_ci_low":
            float(
                ci["lower"]
            ),

        "rank1_ci_high":
            float(
                ci["upper"]
            ),

        "genuine_mean":
            float(
                genuine.mean()
            ),

        "impostor_mean":
            float(
                impostor.mean()
            )
    }

    print(
        f"Rank-1      : "
        f"{result['rank1']*100:.2f}%"
    )

    print(
        f"Rank-5      : "
        f"{result['rank5']*100:.2f}%"
    )

    print(
        f"Rank-10     : "
        f"{result['rank10']*100:.2f}%"
    )

    print(
        f"ROC-AUC     : "
        f"{result['roc_auc']:.4f}"
    )

    print(
        f"EER         : "
        f"{result['eer']*100:.2f}%"
    )

    print(
        f"TAR @ FAR 1%: "
        f"{result['tar_far_1pct']*100:.2f}%"
    )

    print(
        f"TAR @ FAR .1%: "
        f"{result['tar_far_0_1pct']*100:.2f}%"
    )

    print(
        "Rank-1 95% CI: "
        f"["
        f"{result['rank1_ci_low']*100:.2f}%, "
        f"{result['rank1_ci_high']*100:.2f}%"
        f"]"
    )

    del model

    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    return (
        result,
        identification,
        verification,
        genuine,
        impostor,
        probe_y
    )
```

```python
# ============================================================
# PER-IDENTITY ERROR ANALYSIS
# ============================================================

seed = 42

labels = seed_eval[
    seed
]["labels"]

predictions = seed_eval[
    seed
]["identification"][
    "predictions"
]


identity_rows = []

for identity in range(
    NUM_CLASSES
):

    mask = (
        labels == identity
    )

    accuracy = np.mean(
        predictions[mask]
        == labels[mask]
    )

    person = (
        identity // 2
    ) + 1

    hand = (
        "L"
        if identity % 2 == 0
        else "R"
    )

    identity_rows.append({

        "class_id":
            identity,

        "person":
            person,

        "hand":
            hand,

        "correct":
            int(
                np.sum(
                    predictions[mask]
                    == labels[mask]
                )
            ),

        "total":
            int(mask.sum()),

        "accuracy":
            float(accuracy)
    })


identity_performance = (
    pd.DataFrame(
        identity_rows
    )
    .sort_values(
        [
            "accuracy",
            "person",
            "hand"
        ]
    )
    .reset_index(drop=True)
)


print(
    "Hardest hand identities:"
)

display(
    identity_performance.head(
        30
    )
)

identity_performance.to_csv(
    BIO_DIR /
    "identity_error_analysis_seed42.csv",
    index=False
)
```

```python
# ============================================================
# LEFT VS RIGHT PERFORMANCE
# ============================================================

left_mask = (
    labels % 2 == 0
)

right_mask = (
    labels % 2 == 1
)

left_acc = np.mean(
    predictions[left_mask]
    == labels[left_mask]
)

right_acc = np.mean(
    predictions[right_mask]
    == labels[right_mask]
)

print(
    f"Left-hand Rank-1 : "
    f"{left_acc*100:.2f}%"
)

print(
    f"Right-hand Rank-1: "
    f"{right_acc*100:.2f}%"
)
```

```python
# ============================================================
# CONFUSION PAIRS
# ============================================================

cm = confusion_matrix(
    labels,
    predictions,
    labels=np.arange(
        NUM_CLASSES
    )
)

confusion_rows = []

for true_id in range(
    NUM_CLASSES
):

    for pred_id in range(
        NUM_CLASSES
    ):

        if (
            true_id == pred_id
            or cm[
                true_id,
                pred_id
            ] == 0
        ):
            continue

        confusion_rows.append({

            "true_class":
                true_id,

            "true_person":
                true_id // 2 + 1,

            "true_hand":
                (
                    "L"
                    if true_id % 2 == 0
                    else "R"
                ),

            "pred_class":
                pred_id,

            "pred_person":
                pred_id // 2 + 1,

            "pred_hand":
                (
                    "L"
                    if pred_id % 2 == 0
                    else "R"
                ),

            "count":
                int(
                    cm[
                        true_id,
                        pred_id
                    ]
                )
        })


confusion_pairs = (
    pd.DataFrame(
        confusion_rows
    )
    .sort_values(
        "count",
        ascending=False
    )
    .reset_index(drop=True)
)


display(
    confusion_pairs.head(
        30
    )
)

confusion_pairs.to_csv(
    BIO_DIR /
    "confusion_pairs_seed42.csv",
    index=False
)
```

```python
# ============================================================
# OPPOSITE-HAND CONFUSION ANALYSIS
# ============================================================

errors = (
    predictions != labels
)

true_error = labels[
    errors
]

pred_error = predictions[
    errors
]


same_person_opposite_hand = 0

for true_id, pred_id in zip(
    true_error,
    pred_error
):

    true_person = (
        true_id // 2
    )

    pred_person = (
        pred_id // 2
    )

    different_hand = (
        true_id % 2
        != pred_id % 2
    )

    if (
        true_person
        == pred_person
        and different_hand
    ):
        same_person_opposite_hand += 1


total_errors = len(
    true_error
)

rate = (
    same_person_opposite_hand
    / total_errors
    if total_errors > 0
    else 0
)


print(
    "Total Rank-1 errors:",
    total_errors
)

print(
    "Same-person opposite-hand errors:",
    same_person_opposite_hand
)

print(
    "Percentage of errors involving "
    "the opposite hand of the same person:"
)

print(
    f"{rate*100:.2f}%"
)
```
